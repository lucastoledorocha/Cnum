# Atividade 1

Faça na linguagem Python uma função que calcula as raízes da equação do segundo grau.

Serão passados os coeficientes a, b e c. Caso a = 0, resolva como uma equação do primeiro grau, e se b = 0 gere uma exceção.

# Importando uma biblioteca

Código usado para importar uma biblioteca específica em Python.

```python
from math import sqrt
```

# Crinado uma função

Raiz de equação quadrática.

```python
def zeros(a, b, c):

    if a == 0: 
        if b == 0:
            raise ValueError("a e b não podem ser ambos zero, neste caso a equação não é de segundo grau")

        else: 
            return [-c / b]

    if b == 0:
        raise ValueError("b não pode ser zero")

    if a == 0:
        x = -c / b
        return [x]

    delta = b**2 - 4*a*c
    if delta < 0:
        return None
    elif delta == 0:
        x = -b / (2*a)
        return [x, x]
    else:
        x1 = (-b + sqrt(delta)) / (2*a)
        x2 = (-b - sqrt(delta)) / (2*a)
        return [x1, x2]
```

# Executando o código

Criando a função principal.

```python
def main():
    x = zeros (1, -5, 6)
    print(x)
    
if __name__ == "__main__":
    main()
```
