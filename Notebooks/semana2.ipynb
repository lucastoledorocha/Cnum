## Atividade 1
Faça na linguagem Python uma função que verifica se um número é perfeito.

Um número é perfeito se a soma dos divisores for igual a ele mesmo (ex: 6 = 1 + 2 + 3).

````python
def e_perfeito(n: int) -> bool:
    if n < 1:
        return False

    soma_div = 0
    for i in range(1, n):
        if n % i == 0:
            soma_div += i

    return soma_div == n


def main():
    assert e_perfeito(6) == True
    assert e_perfeito(7) == False
    assert e_perfeito(-1) == False


if __name__ == "__main__":
    main()
````

## Atividade 2
Faça na linguagem Python uma função que calcula o fatorial de um número.

O fatorial de um número é o produto do número por todos os antecessores positivos.

````python
def fatorial(n: int) -> int:
    if n < 0:
        raise ValueError("O número deve ser não negativo.")

    result = 1
    i = n
    while i > 1:
        result *= i
        i -= 1

    return result

def main():
    assert fatorial(5) == 120
    assert fatorial(0) == 1
    try:
        fatorial(-1)
    except ValueError as error:
        assert str(error) == "O número deve ser não negativo."

if __name__ == "__main__":
    main()
````

## Atividade 3
Faça na linguagem Python uma função que verifica se um número é primo.

Um número que só pode ser dividido por um e por ele mesmo.

````python
def e_primo(n: int) -> bool:
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0:
        return False

    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2

    return True


def main():
    print(e_primo(1))   
    print(e_primo(3))  
    print(e_primo(15))  
    print(e_primo(18))  
    print(e_primo(2))   


if __name__ == "__main__":
    main()
````

## Atividade 4
Faça na linguagem Python uma função que calcula a soma dos dígitos.

Caso número negativo, dispare uma mensagem de erro (ex: 123, 1 + 2 + 3 = 6).

````python
def soma_dig(n: int) -> int:
    if n < 0:
        raise ValueError("O número deve ser positivo.")

    total = 0
    for digit in str(n):
        total += int(digit)

    return total


def main():
    print(soma_dig(123))
    print(soma_dig(0))
    print(soma_dig(4567))
    try:
        print(soma_dig(-6))
    except ValueError as error:
        print(error)


if __name__ == "__main__":
    main()
````
