## Atividade 1
Escreva os métodos de ponto fixo dentro do arquivo $utils/algoritmos.py$.

````python
import numpy as np
from scipy.optimize import approx_fprime

def pontofixo(a, g, TOL=1e-8):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

def newton_raphson(a, f, TOL=1e-8, df=None):
    if df is None:
        def dfn(x):
            return approx_fprime(np.array([x]), lambda v: f(v[0]))[0]
    else:
        dfn = df
    g = lambda x: x - f(x) / dfn(x)
    return pontofixo(a, g, TOL)

def secante(a, b, f, TOL=1e-8):
    g = lambda a, b: (a * f(b) - b * f(a)) / (f(b) - f(a))
    x = g(a, b)
    while abs(x - b) > TOL:
        a, b = b, x
        x = g(a, b)
    return x
````

Resolver a equação $e^x = x + 2$ é equivalente a calcular os pontos fixos da função  

$$
g(x) = e^x - 2
$$

Use os métodos do ponto fixo  

$$
x^{(n+1)} = g(x^{(n)})
$$

com $x^{(0)} = -1.8$ para obter uma aproximação de uma das soluções da equação dada com 8 dígitos significativos.  

**Resposta:**  
$
x \approx -1.8414057
$

````python
import numpy as np

import sys
import os

# Sobe um nível para a raiz do projeto e adiciona ao caminho do sistema
sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import (
    pontofixo,
    newton_raphson,
    secante,
)

f1 = lambda x: np.e**x - x - 2
g1 = lambda x: np.e**x - 2


def main():
    r = pontofixo(-1.8, g1)
    print(f"raiz ponto fixo = {r}")
    r = newton_raphson(-1.8, f1, df=lambda x: np.e**x - 1)
    print(f"raiz newton-raphson df = {r}")
    r = newton_raphson(-1.8, f1)
    print(f"raiz newton-raphson = {r}")
    r = secante(-1.8, -1.7, f1)
    print(f"raiz secante = {r}")


if __name__ == "__main__":
    main()
````

## Atividade 2
Encontre a raiz positiva da função  

$$
f(x) = \cos(x) - x^2
$$  

pelos métodos do ponto fixo, inicializando-o com $x^{(0)} = 1$.  

Realize a iteração até obter estabilidade no **quinto dígito significativo**.  

**Resposta:**

$$
x \approx 0.82413 
$$

Processo iterativo:  

$$
x^{(n+1)} = x^{(n)} + \frac{\cos(x) - x^2}{\sin(x) + 2x}
$$

````python
import numpy as np



f = lambda x: np.cos(x) - x**2

df = lambda x: -np.sin(x) - 2 * x



x = 1.0

print(f"x^(0) = {x:.10f}")



for n in range(1, 20):

    x_next = x - f(x) / df(x)

    print(f"x^({n}) = {x_next:.10f}")

    if f"{x_next:.5g}" == f"{x:.5g}":

        break

    x = x_next



print(f"\nRaiz aproximada (5 algarismos significativos): {x_next:.5g}")
````

## Atividade 3
Aplique os métodos do ponto fixo para resolver a equação:

$$
e^{-x^2} = 2x
$$

**Resposta:**

$$
x \approx 0.4193648
$$

````python
import numpy as np



g = lambda x: 0.5 * np.exp(-x**2)

x = 0.5

TOL = 1e-8



print(f"x^(0) = {x:.10f}")

for n in range(1, 100):

    x_next = g(x)

    print(f"x^({n}) = {x_next:.10f}")

    if abs(x_next - x) < TOL:

        break

    x = x_next



print(f"\nRaiz aproximada: {x_next:.7f}")
````

## Atividade 4

Resolva os três últimos exercícios da semana anterior usando os métodos do ponto fixo.

````python
import math





def newton_raphson(funcao, derivada, x0, tol=1e-12, max_iter=100):

    x = x0

    for iteracao in range(1, max_iter + 1):

        x_novo = x - funcao(x) / derivada(x)

        if abs(x_novo - x) < tol:

            return x_novo, iteracao

        x = x_novo

    raise RuntimeError("O método de Newton-Raphson não convergiu.")





# Atividade 4: tensão no diodo para cada par (V, R).

IR = 1e-12

k = 1.380649e-23

q = 1.602176634e-19

temperatura = 300.0

vt = k * temperatura / q



def equacao_diodo(vd, tensao, resistencia):

    corrente = IR * math.expm1(vd / vt)

    return resistencia * corrente + vd - tensao



def derivada_diodo(vd, resistencia):

    return resistencia * IR * math.exp(vd / vt) / vt + 1



casos_diodo = [

    (30.0, 1e3), (3.0, 1e3), (3.0, 1e4),

    (0.3, 1e3), (-0.3, 1e3), (-30.0, 1e3), (-30.0, 1e4),

]



print("Atividade 4: tensão no diodo")

for tensao, resistencia in casos_diodo:

    vd, iteracoes = newton_raphson(

        lambda valor: equacao_diodo(valor, tensao, resistencia),

        lambda valor: derivada_diodo(valor, resistencia),

        min(tensao, 0.6),

    )

    print(f"V={tensao:g} V, R={resistencia:g} ohm: vd={vd:.3g} V ({iteracoes} iterações)")





# Atividade 5: comprimento do cabo para d=500 m e flecha máxima de 50 m.

distancia = 500.0

flecha_maxima = 50.0



def equacao_catenaria(comprimento):

    u = distancia / (2 * comprimento)

    return comprimento * (math.cosh(u) - 1) - flecha_maxima



def derivada_catenaria(comprimento):

    u = distancia / (2 * comprimento)

    return math.cosh(u) - 1 - u * math.sinh(u)



comprimento, iteracoes_cabo = newton_raphson(

    equacao_catenaria, derivada_catenaria, 500.0

)

print(f"\nAtividade 5: comprimento do cabo C={comprimento:.4f} m ({iteracoes_cabo} iterações)")





# Atividade 6: ângulo de extinção da corrente no retificador meia onda.

frequencia = 1e3

indutancia = 100e-3

resistencia = 1e3

tan_phi = 2 * math.pi * frequencia * indutancia / resistencia

phi = math.atan(tan_phi)



def corrente_diodo(beta):

    return math.sin(beta - phi) + math.sin(phi) * math.exp(-beta / tan_phi)



def derivada_corrente(beta):

    termo_exponencial = math.exp(-beta / tan_phi)

    return math.cos(beta - phi) - (math.sin(phi) / tan_phi) * termo_exponencial



beta, iteracoes_beta = newton_raphson(

    corrente_diodo, derivada_corrente, 1.5 * math.pi

)

print(f"Atividade 6: beta={math.degrees(beta):.4f} graus ({iteracoes_beta} iterações)")
````
